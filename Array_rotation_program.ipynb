```python
def getMaxElementIndexes(a, rotate):
    n = len(a)

    # Find index of maximum element in original array
    max_val = a[0]
    max_index = 0
    for i in range(1, n):
        if a[i] > max_val:
            max_val = a[i]
            max_index = i

    result = []

    for k in rotate:
        k = k % n
        new_index = (max_index - k) % n
        result.append(new_index)

    return result
```

```python
a = [1, 2, 3]
rotate = [1, 2, 3, 4]

print(getMaxElementIndexes(a, rotate))
```
