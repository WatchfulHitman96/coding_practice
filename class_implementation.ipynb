```python
class PersonAccount:
    def __init__(self, firstname, lastname):
        """
        The constructor method initializes the object.
        """
        self.firstname = firstname
        self.lastname = lastname
        # We use lists to store dictionaries of incomes and expenses
        self.incomes = []
        self.expenses = []

    def add_income(self, description, amount):
        """
        Adds an income entry with a description and amount.
        """
        new_income = {'description': description, 'amount': amount}
        self.incomes.append(new_income)

    def add_expense(self, description, amount):
        """
        Adds an expense entry with a description and amount.
        """
        new_expense = {'description': description, 'amount': amount}
        self.expenses.append(new_expense)

    def total_income(self):
        """
        Calculates the sum of all income amounts.
        """
        total = 0
        for item in self.incomes:
            total += item['amount']
        return total

    def total_expense(self):
        """
        Calculates the sum of all expense amounts.
        """
        total = 0
        for item in self.expenses:
            total += item['amount']
        return total

    def account_balance(self):
        """
        Returns total_income minus total_expense.
        """
        return self.total_income() - self.total_expense()

    def account_info(self):
        """
        Returns a formatted string with the account details.
        """
        info = f"""
        --- Account Info ---
        Name: {self.firstname} {self.lastname}
        Total Income: {self.total_income()}
        Total Expense: {self.total_expense()}
        Current Balance: {self.account_balance()}
        --------------------
        """
        return info

# --- usage Example ---

# 1. Create an instance of the class
person = PersonAccount("Jane", "Doe")

# 2. Add some incomes
person.add_income("Salary", 4000)
person.add_income("Bonus", 500)

# 3. Add some expenses
person.add_expense("Rent", 1200)
person.add_expense("Groceries", 300)
person.add_expense("Utilities", 150)

# 4. Display the info
print(person.account_info())

# 5. Check specific totals individually
print(f"Direct check of balance: {person.account_balance()}")
```
