```python
def get_character_frequency(input_string):
    """
    Counts the frequency of characters in a comma-separated string
    and returns a formatted string with characters and their counts.
    """
    # Step 1: Split the string by commas to get individual elements
    char_list = input_string.split(',')

    # Step 2: Count occurrences using a dictionary
    # This maintains the order of first appearance
    counts = {}
    for char in char_list:
        # If char exists, increment; otherwise, initialize to 1
        counts[char] = counts.get(char, 0) + 1

    # Step 3: Build the output string
    # We join each key (character) and value (count) together
    result = "".join([f"{char}{count}" for char, count in counts.items()])

    return result

# Example Usage:
input_str = "a,a,b,c,b,a,c,d"
output = get_character_frequency(input_str)

print(f"Input: {input_str}")
print(f"Output: {output}")
```
